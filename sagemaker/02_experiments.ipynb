```python
import sagemaker
import boto3
from sagemaker.experiments.run import Run
from sagemaker.sklearn.estimator import SKLearn

# Nombre del experimento
experiment_name = "alquiler-madrid-precio"

# Lanzar training job asociado al experimento
with Run(
    experiment_name=experiment_name,
    run_name="run-gradientboosting-v1",
    sagemaker_session=sagemaker.Session()
) as run:
    
    estimator = SKLearn(
        entry_point="train.py",
        role="arn:aws:iam::<ACCOUNT_ID>:role/LabRole",
        instance_type="ml.m5.large",
        instance_count=1,
        framework_version="1.2-1",
        py_version="py3",
        hyperparameters={
            "n-estimators":  100,
            "max-depth":     5,
            "learning-rate": 0.1
        },
        output_path="s3://pablo-proyectoentrega4/models/",
        base_job_name="alquiler-madrid-regresion",
        enable_sagemaker_metrics=True
    )
    
    estimator.fit({"train": "s3://pablo-proyectoentrega4/processed/"}, wait=True)

print("Experimento registrado correctamente.")
```
